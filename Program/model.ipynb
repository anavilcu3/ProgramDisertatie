```python
import pandas as pd
import pickle
from sklearn.preprocessing import LabelEncoder
from sklearn.ensemble import RandomForestClassifier

categorical_cols = ['Gender', 'Married', 'Dependents', 'Education', 'Self_Employed', 'Property_Area', 'Credit_History']
numeric_cols = ['ApplicantIncome', 'CoapplicantIncome', 'LoanAmount', 'Loan_Amount_Term']

encoders = {}
data = pd.read_csv("dataset/LoanApprovalPrediction.csv")

# 1. Encodăm coloanele categorice (Input Features)
for col in categorical_cols:
    le = LabelEncoder()
    data[col] = le.fit_transform(data[col].astype(str))
    encoders[col] = le

# 2. Encodăm variabila Target (Loan_Status)
target_le = LabelEncoder()
data['Loan_Status'] = target_le.fit_transform(data['Loan_Status'])
encoders['target'] = target_le

# 3. Definim X și Y
X = data[categorical_cols + numeric_cols]
Y = data['Loan_Status']

# 4. Antrenăm modelul
rfc = RandomForestClassifier(n_estimators=100, max_depth=7) # Parametri recomandați pentru stabilitate
rfc.fit(X, Y)

# 5. Exportăm tot ce este necesar pentru predicții viitoare
model_export = {
    "model": rfc,
    "encoders": encoders,
    "features": X.columns.tolist()
}

with open('model_final.pkl', 'wb') as f:
    pickle.dump(model_export, f)
    print("Modelul a fost antrenat și exportat cu succes în 'model_final.pkl'.")
```
